# 03 — Exploratory Data Analysis

This notebook explores patterns in final grades and a few variables that may be related to student performance.


```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("../data/student-mat.csv", sep=";")
```

## Distribution of final grade


```python
sns.histplot(df["G3"], bins=10, kde=True)
plt.title("Distribution of Final Grade (G3)")
plt.xlabel("Final Grade (G3)")
plt.ylabel("Number of Students")
plt.show()
```

## Study time and gender


```python
sns.histplot(data=df, x="studytime", hue="sex", multiple="dodge")
plt.title("Study Time by Gender")
plt.xlabel("Study Time Category")
plt.ylabel("Number of Students")
plt.show()
```

## Study time vs final grade


```python
sns.scatterplot(data=df, x="studytime", y="G3")
plt.title("Study Time vs Final Grade")
plt.xlabel("Study Time Category")
plt.ylabel("Final Grade (G3)")
plt.show()
```

## Average final grade by study time


```python
studytime_mean = df.groupby("studytime", as_index=False)["G3"].mean()

sns.barplot(data=studytime_mean, x="studytime", y="G3")
plt.title("Average Final Grade by Study Time")
plt.xlabel("Study Time Category")
plt.ylabel("Average Final Grade (G3)")
plt.show()
```

## Study time vs final grade distribution


```python
sns.boxplot(data=df, x="studytime", y="G3")
plt.xlabel("Study Time Category")
plt.ylabel("Final Grade (G3)")
plt.title("Study Time vs Final Grade")
plt.show()
```

## Previous failures vs final grade


```python
sns.boxplot(data=df, x="failures", y="G3")
plt.xlabel("Number of Previous Failures")
plt.ylabel("Final Grade (G3)")
plt.title("Previous Failures vs Final Grade")
plt.show()
```

### Takeaway

The exploratory analysis suggests that previous academic difficulty and study-related factors are associated with final grades. These observations help guide the modelling stage, but correlation or visual patterns alone do not establish causation.
