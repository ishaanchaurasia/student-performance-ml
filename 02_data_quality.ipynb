# 02 — Data Quality

Before building a model, check whether the dataset contains missing values or duplicate rows and inspect the target distribution.


```python
import pandas as pd

df = pd.read_csv("../data/student-mat.csv", sep=";")
```

## Missing values


```python
df.isnull().sum()
```

There are no missing values in the dataset.


## Duplicate rows


```python
df.duplicated().sum()
```

There are no duplicate rows in the dataset.


## Final grade distribution


```python
df["G3"].value_counts().sort_index()
```

The final grade is not evenly distributed. In particular, the value **0 occurs relatively often**, while some high grades are much less common. This imbalance is relevant when interpreting predictions for extreme outcomes.


## Basic target statistics


```python
print("Mean:", df["G3"].mean())
print("Median:", df["G3"].median())
print("Mode:", df["G3"].mode()[0])
```

## Study time distribution


```python
df["studytime"].value_counts().sort_index()
```

### Takeaway

The dataset is complete and contains no duplicate rows. However, the distribution of final grades is uneven, which is one reason the model may have more difficulty with unusual or extreme grades.
